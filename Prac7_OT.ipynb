```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.optimize import curve_fit

# Dataset
x = np.array([1, 2, 3, 4, 5])
y = np.array([2.1, 4.2, 5.8, 8.5, 10.1])

# Model Function
def model(x, a, b):
    return a * x + b

# Curve Fitting
params, covariance = curve_fit(model, x, y)

a, b = params

print("Estimated Parameters:")
print("a =", round(a, 4))
print("b =", round(b, 4))

# Predicted Values
y_pred = model(x, a, b)

# Plot Graph
plt.scatter(x, y, label="Actual Data")
plt.plot(x, y_pred, label="Best Fit Curve")
plt.xlabel("X")
plt.ylabel("Y")
plt.title("Least Squares Curve Fitting")
plt.legend()
plt.grid(True)
plt.show()
```
